```python
my_run = BuildStockQuery('ghpphase2','buildstock_sdr',
                         ("comstock_amy2018_r2_2025_md_agg_national_vu",
                          "comstock_amy2018_r2_2025_ts_by_state_vu",
                          "comstock_amy2018_r2_2025_md_agg_national_vu"),
                        buildstock_type='comstock',
                        db_schema="comstock_oedi",
                        skip_reports=False)
```

```python
hvac_columns = [
"out.electricity.heating.energy_consumption",
"out.electricity.cooling.energy_consumption",
"out.electricity.fans.energy_consumption",
"out.electricity.heat_recovery.energy_consumption",
"out.electricity.heat_rejection.energy_consumption",
"out.electricity.pumps.energy_consumption"
]

non_hvac_columns = [
    "out.electricity.exterior_lighting.energy_consumption",
    "out.electricity.interior_lighting.energy_consumption",
    "out.electricity.interior_equipment.energy_consumption",
    "out.electricity.refrigeration.energy_consumption",
    "out.electricity.water_systems.energy_consumption",
]
#computed_col_expr = "out.electricity.total.energy_consumption - " + " - ".join(hvac_columns)
computed_col_expr = " + ".join(non_hvac_columns)
non_hvac_load = my_run.get_calculated_column("out.electricity.non_hvac_load.energy_consumption", computed_col_expr,
                                             table='timeseries')

# for upgrade in [0, 1, 14, 55, 59]:

sqft_col = my_run._get_enduse_cols(["in.sqft"], table="baseline")[0]


for upgrade in [1]:
    print(f"Running for upgrade {upgrade}...")
    agg_df = my_run.query(enduses=[sqft_col, non_hvac_load], annual_only=False,
                                timestamp_grouping_func="hour", group_by=["state"],
                                get_query_only=False, upgrade_id=upgrade)
    agg_df.sort_values(["state", "timestamp"], inplace=True)
    agg_df.to_csv(f"comstock_amy2018_r2_2025_ts_aggregates_upgrade_{upgrade}_2026mar17_directsum.csv", index=False)

```
