```python
my_run = BuildStockQuery('ghpphase2','buildstock_sdr',
                         ("resstock_amy2018_r1_2025_md_national_vu",
                          "resstock_amy2018_r1_2025_ts_by_state_vu",
                          "resstock_amy2018_r1_2025_md_national_vu"),
                        buildstock_type='resstock',
                        db_schema="resstock_oedi",
                        skip_reports=False)
```

```python
hvac_columns = [
"out.electricity.heating.energy_consumption",
"out.electricity.heating_fans_pumps.energy_consumption",
"out.electricity.heating_hp_bkup.energy_consumption",
"out.electricity.heating_hp_bkup_fa.energy_consumption",
"out.electricity.cooling.energy_consumption",
"out.electricity.cooling_fans_pumps.energy_consumption"
]
excluded_columns = [*hvac_columns, "out.electricity.ev_charging.energy_consumption"]
computed_col_expr = "out.electricity.total.energy_consumption - " + " - ".join(excluded_columns)
computed_col_expr
```

```python
hvac_columns = [
"out.electricity.heating.energy_consumption",
"out.electricity.heating_fans_pumps.energy_consumption",
"out.electricity.heating_hp_bkup.energy_consumption",
"out.electricity.heating_hp_bkup_fa.energy_consumption",
"out.electricity.cooling.energy_consumption",
"out.electricity.cooling_fans_pumps.energy_consumption"
]
excluded_columns = [*hvac_columns, "out.electricity.ev_charging.energy_consumption"]
computed_col_expr = "out.electricity.total.energy_consumption - " + " - ".join(excluded_columns)
all_except_pv_and_ev = my_run.get_calculated_column("out.electricity.non_hvac_load.energy_consumption", computed_col_expr,
table='timeseries')
for upgrade in [0, 4, 8]:

    print(f"Running for upgrade {upgrade}...")
    agg_df = my_run.query(enduses=[all_except_pv_and_ev], annual_only=False,
                                timestamp_grouping_func="hour", group_by=["state"],
                                get_query_only=False, upgrade_id=upgrade)
    agg_df.sort_values(["state", "timestamp"], inplace=True)
    agg_df.to_csv(f"resstock_amy2018_r1_2025_upgrade_{upgrade}_2026mar11.csv", index=False)

```
